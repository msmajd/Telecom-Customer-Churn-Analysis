```python
df[ 'Total Charges']=pd.to_numeric(df[ 'Total Charges'],errors='coerce')
df[ 'Total Charges'].isna().sum
```

```python
df[ 'Total Charges'].fillna(0,inplace=True)
```

```python

df[ 'Churn'] = df[ 'Churn Label'].replace({'yes': 1, 'no': 0})

```

```python
df.columns
```

```python
leakage_cols=[
    'CustomerID',
    'CLTV',
    'Churn Score',
    'Churn Reason'
]
df.drop(columns=[col for col in leakage_cols if col in df.columns], inplace=True)
```

```python
def tenure_group(Tenure_Months):
    if Tenure_Months<=6:
        return '0-6 Months'
    elif Tenure_Months<=12:
        return '6-12 Months'
    elif Tenure_Months<=24:
        return '1-2 Years'
    else :
        return '2+ Years'
df['TenureGroup']=df['Tenure Months'].apply(tenure_group)
```
